## 8.지도학습 흐름

```python
긍정단어 = {'좋아', '재미있었음', '잘생김', '예뻐', '연기훌륭함'}
부정단어 = {'지루해', '재미없었음', '비추천'}
```

```python
txt0 = "남주가 너무 잘생김 여주도 예뻐 영화가 중간에 조금 지루해 그래도 아무튼 재미있었음"
```

```python
긍정단어수 = sum([l in 긍정단어 for l in txt0.split(' ')])
부정단어수 = sum([l in 부정단어 for l in txt0.split(' ')])
긍정평가일확률 = 긍정단어수/(긍정단어수+부정단어수)
부정평가일확률 = 부정단어수/(긍정단어수+부정단어수)
긍정평가일확률,부정평가일확률
```

```python
def f(text):
    긍정단어수 = sum([l in 긍정단어 for l in txt0.split(' ')])
    부정단어수 = sum([l in 부정단어 for l in txt0.split(' ')])
    긍정평가일확률 = 긍정단어수/(긍정단어수+부정단어수)
    부정평가일확률 = 부정단어수/(긍정단어수+부정단어수)
    if 긍정평가일확률 > 0.5:
        return 긍정평가일확률, '긍정평가'
    else:
        return 부정평가일확률, '부정평가'
```

```python
f("남주가 너무 잘생김. 여주도 예뻐. 영화가 중간에 조금 지루해 그래도 아무튼 재미있었음.")
```

```python
from datasets import load_dataset
imdb = load_dataset('imdb')
```

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased")
```

```python
dir(tokenizer)
```

```python
tokenizer("Transformers are really useful for natural language processing.") # 텍스트 -> 숫자들
```

```python
def preprocess_function(examples):
    return tokenizer(examples["text"], truncation=True)
```

```python
tokenizer("Transformers are really useful for natural language processing.",truncation=True) # 텍스트 -> 숫자들
```

```python
examples = imdb['train'][1]
preprocess_function(examples)
```

```python
examples = imdb['train'][:2]
preprocess_function(examples)
```

- map 메소드 사용

```python
tokenized_imdb = imdb.map(preprocess_function, batched=True)
```

```python
for i in range(5):
    print(preprocess_function(imdb['train'][i]))
```
